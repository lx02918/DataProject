```python
# 小红的有根树路径选择
import sys
sys.setrecursionlimit(1000000)

def dfs(u, parent, k):
    sum_val = 0
    for v in tree[u]:
        if v != parent:
            sum_val += dfs(v, u, k)
    if a[u] <= k:
        sum_val += a[u]
    else:
        sum_val = float('-inf')
    return max(0, sum_val)
def check(k):
    return dfs(1, 0, k) >= m
n, m = map(int, input().split())
a = [0] + list(map(int, input().split()))
tree = [[] for _ in range(n+1)]

for i in range(1, n + 1):
    children = list(map(int, input().split()))
    for child in children[1:]:
        tree[i].append(child)
        tree[child].append(i)
        
left, right = 1, 10 ** 9
ans = -1
while left <= right:
    mid = (left + right) // 2
    if check(mid):
        ans = mid
        right = mid - 1
    else:
        left = mid + 1
print(ans)
```
