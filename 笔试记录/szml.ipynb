```python
SELECT 
    exam_id,
    CASE
        WHEN tag LIKE '%,%' THEN SUBSTRING_INDEX(tag, ',', 1)
        ELSE tag
    END AS tag,
    CASE
        WHEN difficulty NOT IN ('hard', 'medium', 'easy') AND tag LIKE '%,%' THEN SUBSTRING_INDEX(SUBSTRING_INDEX(tag, ',', 2), ', ', -1)
        ELSE difficulty
    END AS difficulty,
    CASE
        WHEN duration = 0 AND tag LIKE '%,%' THEN CAST(SUBTRING_INDEX(tag, ',', -1) AS UNSIGNED)
        ELSE duration
    END AS duration
FROM examination _info
WHERE exam_id = 9004;
```

```python
class Solution:
    def dnd(self , mp: List[List[int]]) -> int:
        m, n = len(mp), len(mp[0])
        dp = [[float('inf')] * (n + 1) for _ in range(m + 1)]
        dp[m][n - 1] = dp[m - 1][n] = 1
        
        for i in range(m - 1, -1, -1):
            for j in range(n - 1, -1, -1):
                dp[i][j] = max(1, min(dp[i + 1][j], dp[i][j + 1]) - mp[i][j])
        return dp[0][0]
```
