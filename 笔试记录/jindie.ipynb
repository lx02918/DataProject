```python
class Solution:
    def search(self , nums: List[int], target: int) -> int:
        low, high = 0, len(nums) - 1
        while low <= high:
            curr = (high + low) // 2
            if (nums[curr] == target):
                return curr
            elif (target < nums[curr]):
                high = curr - 1
            elif (target > nums[curr]):
                low = curr + 1
        return -1
```

```python
class Solution:
    def search(self , nums: List[int], target: int) -> int:
        i, j = 0, len(nums) - 1
        while i <= j:
            m = (i + j) // 2
            if nums[m] < target: i = m + 1
            elif nums[m] > target: j = m - 1
            else: return m
        return -1
```
